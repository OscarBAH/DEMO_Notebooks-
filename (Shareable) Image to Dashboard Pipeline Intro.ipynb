## Image-to-dashboard workflow

1. **Draft in MS Paint (or similar).**
   * Open MS Paint and sketch a very simple dashboard layout using rectangles, text labels, and chart placeholders.
   * Add titles, subtitles, legends, and any explanatory context you want the final dashboard to show.

2. **Screenshot and send to ChatGPT.**
   * Take a screenshot of the sketch.
   * In ChatGPT, paste the screenshot and ask it to "turn this into a visually polished, responsive HTML dashboard".
   * Copy the full HTML that ChatGPT returns.

3. **Paste HTML into this notebook.**
   * Go to the **"Paste and render HTML dashboard"** cell below.
   * Replace the placeholder HTML inside `html_dashboard = """..."""` with the HTML from ChatGPT.
   * Run the cell to view the dashboard directly in Databricks.

4. **(Optional) Save as an HTML file.**
   * Save the HTML to a file on your machine (e.g. `my_dashboard.html`) or upload it to DBFS / a Unity Catalog volume.
   * Use `show_html_dashboard_from_file(path)` to render it from a file path instead of pasting the whole string.

## Viewing the HTML as a "webapp"

There are two main ways to view the generated dashboard:

* **Inline in this notebook** – the recommended way for quick iteration.
  * Paste the HTML into the `html_dashboard` string and run the cell to see the result immediately.

* **From an HTML file** – closer to a standalone webapp experience.
  * Save the HTML to a file and upload it to:
    * `dbfs:/FileStore/dashboards/` (or another DBFS path), or
    * a Unity Catalog volume (e.g. `/Volumes/<catalog>/<schema>/<volume>/my_dashboard.html`).
  * Call `show_html_dashboard_from_file("<path>")` to render the same dashboard inside this notebook.
